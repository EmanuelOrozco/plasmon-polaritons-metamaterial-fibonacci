# Validación
Recurrencia de la semitraza frente al producto de matrices.

```python
import sys
from pathlib import Path
import numpy as np
sys.path.insert(0, str(Path.cwd().parent / "src"))
from fibonacci_photonics.config import spec_from_mapping
from fibonacci_photonics.physics import Polarization, omega_from_nu_ghz
from fibonacci_photonics.solvers import PWESolver, TMMSolver
from fibonacci_photonics.solvers.tmm import semitrace_by_product, semitrace_by_recurrence
spec = spec_from_mapping({"layer_a_thickness_mm": 12, "layer_b_thickness_mm": 12,
    "omega_e_over_2pi_ghz": 3, "omega_m_over_2pi_ghz": 1})
nu = np.array([0.8, 1.5, 2.5])
omega = omega_from_nu_ghz(nu)
r1 = semitrace_by_recurrence(spec, 5, omega, np.pi/6, Polarization.TE)
r2 = semitrace_by_product(spec, 5, omega, np.pi/6, Polarization.TE)
print("recurrencia vs producto:", np.max(np.abs(r1-r2)))
r_tmm = TMMSolver(spec).semitrace(5, np.pi/6, nu, "TE")
r_pwe = PWESolver(spec).semitrace(5, np.pi/6, nu, "TE")
print("TMM vs PWE:", np.max(np.abs(r_tmm-r_pwe)))
```
