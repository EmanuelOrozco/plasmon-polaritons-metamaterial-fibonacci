# Validación
Recurrencia de la semitraza frente al producto de matrices.

```python
import sys
from pathlib import Path
import numpy as np
sys.path.insert(0, str(Path.cwd().parent / "src"))
from fibonacci_tmm.params import spec_from_mapping
from fibonacci_tmm.electromagnetics import Polarization
from fibonacci_tmm.transfer_matrix import semitrace_by_product, semitrace_by_recurrence
from fibonacci_tmm.params import ghz_to_omega
spec = spec_from_mapping({"layer_a_thickness_mm": 12, "layer_b_thickness_mm": 12,
    "omega_e_over_2pi_ghz": 3, "omega_m_over_2pi_ghz": 1})
omega = ghz_to_omega(np.array([0.8, 1.5, 2.5]))
r1 = semitrace_by_recurrence(spec, 5, omega, np.pi/6, Polarization.TE)
r2 = semitrace_by_product(spec, 5, omega, np.pi/6, Polarization.TE)
print(np.max(np.abs(r1-r2)))
```
